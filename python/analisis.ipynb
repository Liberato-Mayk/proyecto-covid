```python
# cambio semanal
((df["Confirmados"] - df["Confirmados_semana_pasada"]) != df["Cambio_1_semana"]).sum()
```

```python
# letalidad
letal = df["Muertes"]/df["Confirmados"]*100
print((abs(letal - df["Muertes_x100_casos"]) > df["Confirmados"]).sum())
```

```python
print(((df["Muertes"] + df["Recuperados"]) > df["Confirmados"]).sum())
```

```python
df = df.drop(columns="check_activos")
```

```python
df
```

```python
df["Pais"] = df["Pais"].str.strip()
df = df.replace([np.inf, -np.inf], np.nan)   # infinitos a nulo
df = df.drop_duplicates()
```

```python
# creamos la columna letalidad
df["Letalidad"] = df["Muertes"] / df["Confirmados"] * 100
```

```python
# creamos la columna tasa de recuperación
df["Tasa_recuperacion"] = df["Recuperados"] / df["Confirmados"] * 100
```

```python
# porcentaje de casos activos
df["Pct_activos"] = df["Activos"] / df["Confirmados"] * 100
```

```python
# participación en el total mundial
df["Pct_mundial_casos"] = df["Confirmados"] / df["Confirmados"].sum() * 100
```

```python
# ranking mundial por casos
df["Ranking_casos"] = df["Confirmados"].rank(ascending=False, method="min").astype(int)
```

```python
# nivel de crecimiento semanal
df["Nivel_crecimiento"] = pd.cut(df["Aumento_pct_1_semana"], bins=[-np.inf, 5, 15, np.inf], labels=["Bajo", "Medio", "Alto"])
```

```python
# nivel de letalidad
df["Nivel_letalidad"] = pd.cut(df["Letalidad"], bins=[-np.inf, 1, 3, 6, np.inf], labels=["Baja", "Media", "Alta", "Muy alta"])
```

```python
# tamaño del brote
df["Tamano_brote"] = pd.qcut(df["Confirmados"], q=4, labels=["Pequeño", "Mediano", "Grande", "Muy grande"])
```

```python
df
```

```python
# top 10 por casos, muertes y letalidad (filtra países con pocos casos para letalidad)
df.nlargest(10, "Confirmados")[["Pais", "Confirmados"]]
df[df["Confirmados"] > 1000].nlargest(10, "Letalidad")[["Pais", "Letalidad"]]
```

```python
import matplotlib.pyplot as plt

df_filtrado = df[df["Confirmados"] > 1000]

plt.figure(figsize=(10, 5))
plt.hist(df_filtrado["Letalidad"].dropna(), bins=15, edgecolor="black")

plt.title("Distribución de la letalidad por COVID-19")
plt.xlabel("Letalidad (%)")
plt.ylabel("Frecuencia de países")
plt.show()
```

```python
# totales por region
region = df.groupby("Region_OMS").agg(
    paises=("Pais", "count"),
    casos=("Confirmados", "sum"),
    muertes=("Muertes", "sum"),
    recuperados=("Recuperados", "sum")
)
region["letalidad"] = region["muertes"] / region["casos"] * 100
```

```python
# grafico de barras de casos confirmados por region
region["casos"].sort_values(ascending=False).plot(
    kind="bar",
    figsize=(10, 5),
    edgecolor="black"
)

plt.title("Casos confirmados de COVID-19 por región de la OMS")
plt.xlabel("Región de la OMS")
plt.ylabel("Total de casos confirmados")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# gráfico de barras de casos muertes por región
region["muertes"].sort_values(ascending=False).plot(kind="bar", figsize=(10, 5), edgecolor="black")

plt.title("Muertes por COVID-19 según región de la OMS")
plt.xlabel("Región de la OMS")
plt.ylabel("Total de muertes")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# gráfico de barras de casos letalidad por región
region["letalidad"].sort_values(ascending=False).plot(
    kind="bar",
    figsize=(10, 5),
    edgecolor="black"
)

plt.title("Letalidad de COVID-19 por región de la OMS")
plt.xlabel("Región de la OMS")
plt.ylabel("Letalidad (%)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
